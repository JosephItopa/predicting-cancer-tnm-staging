```python
import re
import torch
from transformers import AutoTokenizer, AutoModel

# Load Clinical BERT tokenizer
# tokenizer = AutoTokenizer.from_pretrained("emilyalsentzer/Bio_ClinicalBERT")

def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove emojis and special unicode characters
    # This regex removes anything that's not a letter, number, basic punctuation, or whitespace
    text = re.sub(r"[^\w\s.,;:!?'\-]", " ", text)

    # Replace multiple spaces/newlines with a single space
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import torch
from transformers import AutoTokenizer, AutoModel

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ClinicalBERT
MODEL_NAME_CLINICALBERT = "emilyalsentzer/Bio_ClinicalBERT"
tokenizer_clinical = AutoTokenizer.from_pretrained(MODEL_NAME_CLINICALBERT)
model_clinical = AutoModel.from_pretrained(MODEL_NAME_CLINICALBERT).to(device)
model_clinical.eval()
```

```python
# BioBERT
MODEL_NAME_BIOBERT = "dmis-lab/biobert-base-cased-v1.1"
tokenizer_biobert = AutoTokenizer.from_pretrained(MODEL_NAME_BIOBERT)
model_biobert = AutoModel.from_pretrained(MODEL_NAME_BIOBERT).to(device)
model_biobert.eval()
```

```python
# PubMedBERT
MODEL_NAME_PUBMEDBERT = "microsoft/BiomedNLP-PubMedBERT-base-uncased-abstract-fulltext"
tokenizer_pubmedbert = AutoTokenizer.from_pretrained(MODEL_NAME_PUBMEDBERT)
model_pubmedbert = AutoModel.from_pretrained(MODEL_NAME_PUBMEDBERT).to(device)
model_pubmedbert.eval()
```

```python
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output.last_hidden_state
    mask = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * mask, 1) / torch.clamp(mask.sum(1), min=1e-9)


def get_embeddings(
    texts,
    model_name="emilyalsentzer/Bio_ClinicalBERT",
    max_length=512,
    batch_size=16
):
    """
    Generate embeddings for a list of texts using ClinicalBERT or BioBERT.

    Args:
        texts (list[str]): list of input strings
        model_name (str): Hugging Face model name
            Options:
                - "emilyalsentzer/Bio_ClinicalBERT"
                - "dmis-lab/biobert-base-cased-v1.1"
        max_length (int): maximum token length
        batch_size (int): batch size for inference

    Returns:
        torch.Tensor: (len(texts), hidden_dim) embeddings
    """

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name).to(device)
    model.eval()

    cleaned_texts = [clean_text(t) for t in texts]

    embeddings = []

    for i in range(0, len(cleaned_texts), batch_size):
        batch = cleaned_texts[i:i + batch_size]

        encoded = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        )

        encoded = {k: v.to(device) for k, v in encoded.items()}

        with torch.no_grad():
            outputs = model(**encoded)

        pooled = mean_pooling(outputs, encoded["attention_mask"])
        embeddings.append(pooled.cpu())

    return torch.cat(embeddings, dim=0)
```

```python
MODEL_NAME_CLINICALBERT = "emilyalsentzer/Bio_ClinicalBERT"
MODEL_NAME_BIOBERT = "dmis-lab/biobert-base-cased-v1.1"
MODEL_NAME_BIOMED = "microsoft/BiomedNLP-PubMedBERT-base-uncased-abstract-fulltext"
```

```python
# MODEL_NAME_CLINICALBERT = "emilyalsentzer/Bio_ClinicalBERT"
texts_t = train["text"].tolist()

embeddings_clinicalbert = get_embeddings(texts_t, model_name=MODEL_NAME_CLINICALBERT)

train["text_embedding_clinicalbert"] = embeddings_clinicalbert.tolist()
```

```python
#MODEL_NAME_BIOBERT = "dmis-lab/biobert-base-cased-v1.1"
embeddings_biobert = get_embeddings(texts_t, model_name=MODEL_NAME_BIOBERT)

train["text_embedding_biobert"] = embeddings_biobert.tolist()
```

```python
#MODEL_NAME_BIOMED = "microsoft/BiomedNLP-PubMedBERT-base-uncased-abstract-fulltext"
embeddings_pubmedbert = get_embeddings(texts_t, model_name=MODEL_NAME_BIOMED)

train["text_embedding_pubmedbert"] = embeddings_pubmedbert.tolist()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

def tfidf_embeddings(text_series, max_features=512):
    vectorizer = TfidfVectorizer(
        ngram_range=(1, 2),
        max_features=max_features,
        stop_words='english'
    )

    tfidf_matrix = vectorizer.fit_transform(text_series.astype(str))

    feature_names = vectorizer.get_feature_names_out()
    df_embeddings = pd.DataFrame(
        tfidf_matrix.toarray(),
        columns=feature_names,
        index=text_series.index
    )

    return df_embeddings, vectorizer
```

```python
tfidf_vectors, vectorizer = tfidf_embeddings(train["text"], max_features=512)

train["tfidf_vectors"] = tfidf_vectors.values.tolist()
```
